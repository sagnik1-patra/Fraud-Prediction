```python
import os
import json
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score,
    roc_curve
)
from sklearn.utils.class_weight import compute_class_weight

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from torchvision.models import efficientnet_b0, EfficientNet_B0_Weights


# =========================
# 1. Paths
# =========================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Fraud Prediction"
DATA_DIR = os.path.join(BASE_DIR, "the-freuid-challenge-2026-ijcai-ecai")

TRAIN_IMG_DIR = os.path.join(DATA_DIR, "train", "train")
TRAIN_SAMPLE_IMG_DIR = os.path.join(DATA_DIR, "train_sample", "train_sample")
TEST_IMG_DIR = os.path.join(DATA_DIR, "public_test", "public_test")

TRAIN_LABELS = os.path.join(DATA_DIR, "train_labels.csv")
TRAIN_SAMPLE_LABELS = os.path.join(DATA_DIR, "train_sample_labels.csv")
SAMPLE_SUBMISSION = os.path.join(DATA_DIR, "sample_submission.csv")

RESULT_DIR = os.path.join(BASE_DIR, "fraud_results_pytorch_effb0")
os.makedirs(RESULT_DIR, exist_ok=True)

PREFIX = "fraud_pytorch_effb0_"

print("Results will be saved in:", RESULT_DIR)


# =========================
# 2. Settings
# =========================

IMG_SIZE = 224
BATCH_SIZE = 16
EPOCHS_STAGE_1 = 5
EPOCHS_STAGE_2 = 8
SEED = 42
THRESHOLD = 0.30

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", DEVICE)


# =========================
# 3. Load CSVs
# =========================

train_df = pd.read_csv(TRAIN_LABELS)
sample_df = pd.read_csv(TRAIN_SAMPLE_LABELS)
submission_df = pd.read_csv(SAMPLE_SUBMISSION)

print("Train labels:", train_df.shape)
print("Sample labels:", sample_df.shape)
print("Submission:", submission_df.shape)

print("Train columns:", train_df.columns.tolist())
print("Submission columns:", submission_df.columns.tolist())


# =========================
# 4. Detect Columns
# =========================

def find_id_column(df):
    possible = ["id", "image_id", "filename", "file_name", "image", "path"]
    for col in possible:
        if col in df.columns:
            return col
    return df.columns[0]


def find_label_column(df):
    possible = ["label", "target", "class", "is_fraud", "fraud"]
    for col in possible:
        if col in df.columns:
            return col
    return df.columns[-1]


ID_COL = find_id_column(train_df)
LABEL_COL = find_label_column(train_df)
SUB_ID_COL = find_id_column(submission_df)

print("ID column:", ID_COL)
print("Label column:", LABEL_COL)
print("Submission ID column:", SUB_ID_COL)


# =========================
# 5. Image Path Finder
# =========================

def find_image_path(image_id, folder):
    image_id = str(image_id).strip()

    possible_names = [
        image_id,
        image_id + ".jpg",
        image_id + ".jpeg",
        image_id + ".png",
        image_id + ".webp",
        image_id + ".bmp"
    ]

    for name in possible_names:
        path = os.path.join(folder, name)
        if os.path.exists(path):
            return path

    for root, dirs, files in os.walk(folder):
        for file in files:
            file_name, ext = os.path.splitext(file)
            if file_name == image_id or file == image_id:
                return os.path.join(root, file)

    return None


train_df["full_path"] = train_df[ID_COL].apply(lambda x: find_image_path(x, TRAIN_IMG_DIR))
sample_df["full_path"] = sample_df[ID_COL].apply(lambda x: find_image_path(x, TRAIN_SAMPLE_IMG_DIR))

train_df = train_df.dropna(subset=["full_path"]).reset_index(drop=True)
sample_df = sample_df.dropna(subset=["full_path"]).reset_index(drop=True)

train_df["label"] = train_df[LABEL_COL].astype(int)
sample_df["label"] = sample_df[LABEL_COL].astype(int)

full_df = pd.concat(
    [
        train_df[[ID_COL, "full_path", "label"]],
        sample_df[[ID_COL, "full_path", "label"]]
    ],
    ignore_index=True
)

print("Valid train images:", len(train_df))
print("Valid sample images:", len(sample_df))
print("Total usable images:", len(full_df))

full_df.to_csv(os.path.join(RESULT_DIR, PREFIX + "final_training_data.csv"), index=False)


# =========================
# 6. Label Distribution Graph
# =========================

label_counts = full_df["label"].value_counts().sort_index()
print(label_counts)

plt.figure(figsize=(6, 4))
label_counts.plot(kind="bar")
plt.title("Fraud Label Distribution")
plt.xlabel("Label")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "label_distribution_graph.png"), dpi=300)
plt.show()


# =========================
# 7. Train Validation Split
# =========================

train_data, val_data = train_test_split(
    full_df,
    test_size=0.2,
    random_state=SEED,
    stratify=full_df["label"]
)

print("Train data:", train_data.shape)
print("Validation data:", val_data.shape)


# =========================
# 8. Class Weights
# =========================

classes = np.unique(full_df["label"])

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=full_df["label"]
)

class_weights = {
    int(classes[i]): float(weights[i])
    for i in range(len(classes))
}

# Extra fraud boost
if 1 in class_weights:
    class_weights[1] = class_weights[1] * 2.0

print("Class weights:", class_weights)

with open(os.path.join(RESULT_DIR, PREFIX + "class_weights.json"), "w") as f:
    json.dump(class_weights, f, indent=4)


pos_weight_value = class_weights.get(1, 1.0) / class_weights.get(0, 1.0)
pos_weight = torch.tensor([pos_weight_value], dtype=torch.float32).to(DEVICE)

print("Positive fraud weight:", pos_weight_value)


# =========================
# 9. Dataset Class
# =========================

class FraudImageDataset(Dataset):
    def __init__(self, dataframe, transform=None, is_test=False):
        self.df = dataframe.reset_index(drop=True)
        self.transform = transform
        self.is_test = is_test

    def __len__(self):
        return len(self.df)

    def __getitem__(self, index):
        img_path = self.df.loc[index, "full_path"]

        image = Image.open(img_path).convert("RGB")

        if self.transform:
            image = self.transform(image)

        if self.is_test:
            return image

        label = float(self.df.loc[index, "label"])
        label = torch.tensor(label, dtype=torch.float32)

        return image, label


# =========================
# 10. Transforms
# =========================

train_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(5),
    transforms.ColorJitter(
        brightness=0.15,
        contrast=0.15,
        saturation=0.10
    ),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

val_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


train_dataset = FraudImageDataset(train_data, transform=train_transform)
val_dataset = FraudImageDataset(val_data, transform=val_transform)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)


# =========================
# 11. Build EfficientNetB0
# =========================

weights = EfficientNet_B0_Weights.DEFAULT
model = efficientnet_b0(weights=weights)

in_features = model.classifier[1].in_features

model.classifier = nn.Sequential(
    nn.Dropout(p=0.35),
    nn.Linear(in_features, 256),
    nn.ReLU(),
    nn.BatchNorm1d(256),
    nn.Dropout(p=0.35),
    nn.Linear(256, 1)
)

model = model.to(DEVICE)

criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.0003
)


# =========================
# 12. Helper Functions
# =========================

def train_one_epoch(model, loader, criterion, optimizer):
    model.train()

    running_loss = 0.0
    all_labels = []
    all_probs = []

    for images, labels in loader:
        images = images.to(DEVICE)
        labels = labels.to(DEVICE).view(-1, 1)

        optimizer.zero_grad()

        logits = model(images)
        loss = criterion(logits, labels)

        loss.backward()
        optimizer.step()

        probs = torch.sigmoid(logits).detach().cpu().numpy().ravel()

        running_loss += loss.item()
        all_labels.extend(labels.detach().cpu().numpy().ravel())
        all_probs.extend(probs)

    avg_loss = running_loss / len(loader)
    all_labels = np.array(all_labels)
    all_probs = np.array(all_probs)
    all_preds = (all_probs >= THRESHOLD).astype(int)

    acc = accuracy_score(all_labels, all_preds)

    try:
        auc = roc_auc_score(all_labels, all_probs)
    except:
        auc = 0.0

    return avg_loss, acc, auc


def validate_one_epoch(model, loader, criterion):
    model.eval()

    running_loss = 0.0
    all_labels = []
    all_probs = []

    with torch.no_grad():
        for images, labels in loader:
            images = images.to(DEVICE)
            labels = labels.to(DEVICE).view(-1, 1)

            logits = model(images)
            loss = criterion(logits, labels)

            probs = torch.sigmoid(logits).cpu().numpy().ravel()

            running_loss += loss.item()
            all_labels.extend(labels.cpu().numpy().ravel())
            all_probs.extend(probs)

    avg_loss = running_loss / len(loader)

    all_labels = np.array(all_labels)
    all_probs = np.array(all_probs)
    all_preds = (all_probs >= THRESHOLD).astype(int)

    acc = accuracy_score(all_labels, all_preds)

    try:
        auc = roc_auc_score(all_labels, all_probs)
    except:
        auc = 0.0

    return avg_loss, acc, auc, all_labels, all_probs


# =========================
# 13. Stage 1 Training
# Freeze feature extractor
# =========================

for param in model.features.parameters():
    param.requires_grad = False

optimizer = torch.optim.Adam(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=0.0005
)

history = {
    "train_loss": [],
    "val_loss": [],
    "train_accuracy": [],
    "val_accuracy": [],
    "train_auc": [],
    "val_auc": []
}

best_auc = 0.0
best_model_path = os.path.join(RESULT_DIR, PREFIX + "best_model.pth")

print("\nStage 1: Training classifier only")

for epoch in range(EPOCHS_STAGE_1):
    train_loss, train_acc, train_auc = train_one_epoch(
        model, train_loader, criterion, optimizer
    )

    val_loss, val_acc, val_auc, val_true, val_probs = validate_one_epoch(
        model, val_loader, criterion
    )

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["train_accuracy"].append(train_acc)
    history["val_accuracy"].append(val_acc)
    history["train_auc"].append(train_auc)
    history["val_auc"].append(val_auc)

    print(
        f"Epoch {epoch+1}/{EPOCHS_STAGE_1} | "
        f"Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f} | "
        f"Train AUC: {train_auc:.4f} | Val AUC: {val_auc:.4f}"
    )

    if val_auc > best_auc:
        best_auc = val_auc
        torch.save(model.state_dict(), best_model_path)


# =========================
# 14. Stage 2 Fine-Tuning
# Unfreeze last layers
# =========================

for param in model.features.parameters():
    param.requires_grad = True

# Freeze earlier feature blocks, train last blocks
for idx, child in enumerate(model.features.children()):
    if idx < 6:
        for param in child.parameters():
            param.requires_grad = False

optimizer = torch.optim.Adam(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=0.00002
)

print("\nStage 2: Fine-tuning EfficientNetB0")

for epoch in range(EPOCHS_STAGE_2):
    train_loss, train_acc, train_auc = train_one_epoch(
        model, train_loader, criterion, optimizer
    )

    val_loss, val_acc, val_auc, val_true, val_probs = validate_one_epoch(
        model, val_loader, criterion
    )

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["train_accuracy"].append(train_acc)
    history["val_accuracy"].append(val_acc)
    history["train_auc"].append(train_auc)
    history["val_auc"].append(val_auc)

    print(
        f"Fine Tune Epoch {epoch+1}/{EPOCHS_STAGE_2} | "
        f"Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f} | "
        f"Train AUC: {train_auc:.4f} | Val AUC: {val_auc:.4f}"
    )

    if val_auc > best_auc:
        best_auc = val_auc
        torch.save(model.state_dict(), best_model_path)


# =========================
# 15. Load Best Model
# =========================

model.load_state_dict(torch.load(best_model_path, map_location=DEVICE))
model.eval()

torch.save(model.state_dict(), os.path.join(RESULT_DIR, PREFIX + "final_model.pth"))

pd.DataFrame(history).to_csv(
    os.path.join(RESULT_DIR, PREFIX + "training_history.csv"),
    index=False
)


# =========================
# 16. Training Graphs
# =========================

def save_graph(train_values, val_values, title, ylabel, filename):
    plt.figure(figsize=(7, 5))
    plt.plot(train_values, label="Train")
    plt.plot(val_values, label="Validation")
    plt.title(title)
    plt.xlabel("Epoch")
    plt.ylabel(ylabel)
    plt.legend()
    plt.tight_layout()
    plt.savefig(os.path.join(RESULT_DIR, filename), dpi=300)
    plt.show()


save_graph(
    history["train_accuracy"],
    history["val_accuracy"],
    "EfficientNetB0 Accuracy",
    "Accuracy",
    PREFIX + "accuracy_graph.png"
)

save_graph(
    history["train_loss"],
    history["val_loss"],
    "EfficientNetB0 Loss",
    "Loss",
    PREFIX + "loss_graph.png"
)

save_graph(
    history["train_auc"],
    history["val_auc"],
    "EfficientNetB0 AUC",
    "AUC",
    PREFIX + "auc_graph.png"
)


# =========================
# 17. Final Validation
# =========================

val_loss, val_acc, val_auc, val_true, val_probs = validate_one_epoch(
    model, val_loader, criterion
)

val_pred = (val_probs >= THRESHOLD).astype(int)

cm = confusion_matrix(val_true, val_pred)
report = classification_report(val_true, val_pred, output_dict=True)
text_report = classification_report(val_true, val_pred)

print("\nFinal Validation Results")
print("Accuracy:", val_acc)
print("AUC:", val_auc)
print(text_report)

score_stats = {
    "min_score": float(np.min(val_probs)),
    "max_score": float(np.max(val_probs)),
    "mean_score": float(np.mean(val_probs)),
    "std_score": float(np.std(val_probs))
}

print("Validation score stats:", score_stats)


metrics = {
    "threshold": THRESHOLD,
    "accuracy": float(val_acc),
    "auc": float(val_auc),
    "confusion_matrix": cm.tolist(),
    "classification_report": report,
    "score_statistics": score_stats,
    "class_weights": class_weights
}

with open(os.path.join(RESULT_DIR, PREFIX + "metrics.json"), "w") as f:
    json.dump(metrics, f, indent=4)

pd.DataFrame(report).transpose().to_csv(
    os.path.join(RESULT_DIR, PREFIX + "classification_report.csv")
)


# =========================
# 18. Confusion Matrix
# =========================

plt.figure(figsize=(5, 4))
plt.imshow(cm)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.colorbar()

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j], ha="center", va="center")

plt.xticks([0, 1], ["Real", "Fraud"])
plt.yticks([0, 1], ["Real", "Fraud"])
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "confusion_matrix_heatmap.png"), dpi=300)
plt.show()


# =========================
# 19. ROC Curve
# =========================

fpr, tpr, thresholds = roc_curve(val_true, val_probs)

plt.figure(figsize=(6, 5))
plt.plot(fpr, tpr, label=f"AUC = {val_auc:.4f}")
plt.plot([0, 1], [0, 1], linestyle="--")
plt.title("ROC Curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "roc_curve.png"), dpi=300)
plt.show()


# =========================
# 20. Score Distribution
# =========================

plt.figure(figsize=(7, 5))
plt.hist(val_probs, bins=40)
plt.title("Validation Fraud Score Distribution")
plt.xlabel("Fraud Score")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "validation_score_distribution.png"), dpi=300)
plt.show()


# =========================
# 21. Validation Predictions CSV
# =========================

val_output = val_data.copy()
val_output["fraud_score"] = val_probs
val_output["predicted_label"] = val_pred
val_output["threshold"] = THRESHOLD

val_output.to_csv(
    os.path.join(RESULT_DIR, PREFIX + "validation_predictions.csv"),
    index=False
)


# =========================
# 22. Prepare Test Data
# =========================

submission_df["full_path"] = submission_df[SUB_ID_COL].apply(
    lambda x: find_image_path(x, TEST_IMG_DIR)
)

print("Missing test images:", submission_df["full_path"].isna().sum())

test_df = submission_df.dropna(subset=["full_path"]).reset_index(drop=True)

test_dataset = FraudImageDataset(
    test_df,
    transform=val_transform,
    is_test=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)


# =========================
# 23. Predict Test
# =========================

test_scores = []

model.eval()

with torch.no_grad():
    for images in test_loader:
        images = images.to(DEVICE)
        logits = model(images)
        probs = torch.sigmoid(logits).cpu().numpy().ravel()
        test_scores.extend(probs)

test_scores = np.array(test_scores)

test_df["label"] = test_scores

final_submission = submission_df[[SUB_ID_COL]].merge(
    test_df[[SUB_ID_COL, "label"]],
    on=SUB_ID_COL,
    how="left"
)

final_submission["label"] = final_submission["label"].fillna(0.0)

submission_path = os.path.join(RESULT_DIR, PREFIX + "submission.csv")
final_submission.to_csv(submission_path, index=False)

print("Submission saved:", submission_path)
print(final_submission.head())


# =========================
# 24. Test Score Stats
# =========================

test_score_stats = {
    "min_score": float(final_submission["label"].min()),
    "max_score": float(final_submission["label"].max()),
    "mean_score": float(final_submission["label"].mean()),
    "std_score": float(final_submission["label"].std())
}

print("Test score stats:", test_score_stats)

with open(os.path.join(RESULT_DIR, PREFIX + "test_score_statistics.json"), "w") as f:
    json.dump(test_score_stats, f, indent=4)


plt.figure(figsize=(7, 5))
plt.hist(final_submission["label"], bins=40)
plt.title("Public Test Fraud Score Distribution")
plt.xlabel("Fraud Score")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "test_score_distribution.png"), dpi=300)
plt.show()


# =========================
# 25. Final Summary
# =========================

summary_path = os.path.join(RESULT_DIR, PREFIX + "result_summary.txt")

with open(summary_path, "w") as f:
    f.write("FREUID Fraud Detection - PyTorch EfficientNetB0\n")
    f.write("==============================================\n\n")
    f.write(f"Device used: {DEVICE}\n")
    f.write(f"Image size: {IMG_SIZE}\n")
    f.write(f"Batch size: {BATCH_SIZE}\n")
    f.write(f"Threshold: {THRESHOLD}\n")
    f.write(f"Best validation AUC: {best_auc:.4f}\n")
    f.write(f"Final validation accuracy: {val_acc:.4f}\n")
    f.write(f"Final validation AUC: {val_auc:.4f}\n\n")
    f.write("Validation score statistics:\n")
    f.write(json.dumps(score_stats, indent=4))
    f.write("\n\nTest score statistics:\n")
    f.write(json.dumps(test_score_stats, indent=4))
    f.write("\n\nGenerated files:\n")
    f.write("- fraud_pytorch_effb0_best_model.pth\n")
    f.write("- fraud_pytorch_effb0_final_model.pth\n")
    f.write("- fraud_pytorch_effb0_submission.csv\n")
    f.write("- fraud_pytorch_effb0_metrics.json\n")
    f.write("- fraud_pytorch_effb0_classification_report.csv\n")
    f.write("- fraud_pytorch_effb0_confusion_matrix_heatmap.png\n")
    f.write("- fraud_pytorch_effb0_roc_curve.png\n")
    f.write("- fraud_pytorch_effb0_auc_graph.png\n")
    f.write("- fraud_pytorch_effb0_test_score_distribution.png\n")

print("All results saved in:", RESULT_DIR)
```
