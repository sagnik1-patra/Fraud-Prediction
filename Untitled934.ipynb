```python
import os
import json
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, classification_report, confusion_matrix,
    roc_auc_score, roc_curve
)

import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau


# =========================
# 1. Paths
# =========================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Fraud Prediction"
DATA_DIR = os.path.join(BASE_DIR, "the-freuid-challenge-2026-ijcai-ecai")

TRAIN_IMG_DIR = os.path.join(DATA_DIR, "train", "train")
TRAIN_SAMPLE_IMG_DIR = os.path.join(DATA_DIR, "train_sample", "train_sample")
TEST_IMG_DIR = os.path.join(DATA_DIR, "public_test", "public_test")

TRAIN_LABELS = os.path.join(DATA_DIR, "train_labels.csv")
TRAIN_SAMPLE_LABELS = os.path.join(DATA_DIR, "train_sample_labels.csv")
SAMPLE_SUBMISSION = os.path.join(DATA_DIR, "sample_submission.csv")

RESULT_DIR = os.path.join(BASE_DIR, "fraud_results")
os.makedirs(RESULT_DIR, exist_ok=True)

PREFIX = "fraud_"

print("Saving results to:", RESULT_DIR)


# =========================
# 2. Settings
# =========================

IMG_SIZE = 224
BATCH_SIZE = 32
EPOCHS = 15
SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)


# =========================
# 3. Load Labels
# =========================

train_df = pd.read_csv(TRAIN_LABELS)
sample_df = pd.read_csv(TRAIN_SAMPLE_LABELS)
sub_df = pd.read_csv(SAMPLE_SUBMISSION)

print("Train labels:", train_df.shape)
print("Train sample labels:", sample_df.shape)
print("Sample submission:", sub_df.shape)

print(train_df.head())


# =========================
# 4. Resolve Image Paths
# =========================

def resolve_train_path(row):
    img_path = str(row["image_path"])

    possible_paths = [
        os.path.join(TRAIN_IMG_DIR, img_path),
        os.path.join(TRAIN_IMG_DIR, row["id"] + ".jpg"),
        os.path.join(TRAIN_IMG_DIR, row["id"] + ".png"),
        os.path.join(TRAIN_IMG_DIR, row["id"] + ".jpeg"),
    ]

    for p in possible_paths:
        if os.path.exists(p):
            return p

    return None


def resolve_sample_path(row):
    img_path = str(row["image_path"])

    possible_paths = [
        os.path.join(TRAIN_SAMPLE_IMG_DIR, img_path),
        os.path.join(TRAIN_SAMPLE_IMG_DIR, row["id"] + ".jpg"),
        os.path.join(TRAIN_SAMPLE_IMG_DIR, row["id"] + ".png"),
        os.path.join(TRAIN_SAMPLE_IMG_DIR, row["id"] + ".jpeg"),
    ]

    for p in possible_paths:
        if os.path.exists(p):
            return p

    return None


train_df["full_path"] = train_df.apply(resolve_train_path, axis=1)
sample_df["full_path"] = sample_df.apply(resolve_sample_path, axis=1)

train_df = train_df.dropna(subset=["full_path"]).reset_index(drop=True)
sample_df = sample_df.dropna(subset=["full_path"]).reset_index(drop=True)

print("Valid train images:", train_df.shape)
print("Valid sample images:", sample_df.shape)

full_df = pd.concat([train_df, sample_df], ignore_index=True)
full_df["label"] = full_df["label"].astype(int)

full_df.to_csv(os.path.join(RESULT_DIR, PREFIX + "final_training_data.csv"), index=False)


# =========================
# 5. Label Distribution Graph
# =========================

label_counts = full_df["label"].value_counts().sort_index()
label_counts.to_csv(os.path.join(RESULT_DIR, PREFIX + "label_distribution.csv"))

plt.figure(figsize=(6, 4))
label_counts.plot(kind="bar")
plt.title("Fraud Label Distribution")
plt.xlabel("Label")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "label_distribution_graph.png"), dpi=300)
plt.show()


# =========================
# 6. Train-Validation Split
# =========================

train_data, val_data = train_test_split(
    full_df,
    test_size=0.2,
    random_state=SEED,
    stratify=full_df["label"]
)

print("Training:", train_data.shape)
print("Validation:", val_data.shape)


# =========================
# 7. TensorFlow Dataset
# =========================

def load_image(path, label=None):
    image = tf.io.read_file(path)
    image = tf.image.decode_image(image, channels=3, expand_animations=False)
    image = tf.image.resize(image, [IMG_SIZE, IMG_SIZE])
    image = image / 255.0

    if label is None:
        return image

    return image, tf.cast(label, tf.float32)


def make_dataset(df, shuffle=True):
    paths = df["full_path"].values
    labels = df["label"].values

    ds = tf.data.Dataset.from_tensor_slices((paths, labels))
    ds = ds.map(load_image, num_parallel_calls=tf.data.AUTOTUNE)

    if shuffle:
        ds = ds.shuffle(buffer_size=1000, seed=SEED)

    ds = ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
    return ds


train_ds = make_dataset(train_data, shuffle=True)
val_ds = make_dataset(val_data, shuffle=False)


# =========================
# 8. CNN Model
# =========================

data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.05),
    layers.RandomZoom(0.1),
    layers.RandomContrast(0.1),
])

base_model = tf.keras.applications.EfficientNetB0(
    include_top=False,
    weights="imagenet",
    input_shape=(IMG_SIZE, IMG_SIZE, 3)
)

base_model.trainable = False

inputs = layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3))
x = data_augmentation(inputs)
x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)
x = layers.Dense(128, activation="relu")(x)
x = layers.Dropout(0.3)(x)
outputs = layers.Dense(1, activation="sigmoid")(x)

model = models.Model(inputs, outputs)

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0005),
    loss="binary_crossentropy",
    metrics=["accuracy", tf.keras.metrics.AUC(name="auc")]
)

model.summary()


# =========================
# 9. Train Model
# =========================

best_model_path = os.path.join(RESULT_DIR, PREFIX + "best_model.keras")

callbacks = [
    EarlyStopping(monitor="val_auc", patience=4, mode="max", restore_best_weights=True),
    ModelCheckpoint(best_model_path, monitor="val_auc", mode="max", save_best_only=True),
    ReduceLROnPlateau(monitor="val_loss", patience=2, factor=0.3)
]

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    callbacks=callbacks
)

model.save(os.path.join(RESULT_DIR, PREFIX + "final_model.keras"))
model.save(os.path.join(RESULT_DIR, PREFIX + "final_model.h5"))


# =========================
# 10. Save Training History
# =========================

history_df = pd.DataFrame(history.history)
history_df.to_csv(os.path.join(RESULT_DIR, PREFIX + "training_history.csv"), index=False)

plt.figure(figsize=(7, 5))
plt.plot(history.history["accuracy"], label="Train Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.title("Training Accuracy Graph")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "training_accuracy_graph.png"), dpi=300)
plt.show()

plt.figure(figsize=(7, 5))
plt.plot(history.history["loss"], label="Train Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("Training Loss Graph")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "training_loss_graph.png"), dpi=300)
plt.show()

plt.figure(figsize=(7, 5))
plt.plot(history.history["auc"], label="Train AUC")
plt.plot(history.history["val_auc"], label="Validation AUC")
plt.title("AUC Graph")
plt.xlabel("Epoch")
plt.ylabel("AUC")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "auc_graph.png"), dpi=300)
plt.show()


# =========================
# 11. Validation Predictions
# =========================

val_probs = model.predict(val_ds).ravel()
val_true = val_data["label"].values
val_pred = (val_probs >= 0.5).astype(int)

acc = accuracy_score(val_true, val_pred)
auc = roc_auc_score(val_true, val_probs)
cm = confusion_matrix(val_true, val_pred)
report = classification_report(val_true, val_pred, output_dict=True)

metrics = {
    "accuracy": float(acc),
    "roc_auc": float(auc),
    "confusion_matrix": cm.tolist(),
    "classification_report": report
}

with open(os.path.join(RESULT_DIR, PREFIX + "metrics.json"), "w") as f:
    json.dump(metrics, f, indent=4)

pd.DataFrame(report).transpose().to_csv(
    os.path.join(RESULT_DIR, PREFIX + "classification_report.csv")
)

print("Accuracy:", acc)
print("ROC AUC:", auc)
print(classification_report(val_true, val_pred))


# =========================
# 12. Confusion Matrix Heatmap
# =========================

plt.figure(figsize=(5, 4))
plt.imshow(cm)
plt.title("Confusion Matrix Heatmap")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.colorbar()

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j], ha="center", va="center")

plt.xticks([0, 1], ["Real", "Fraud"])
plt.yticks([0, 1], ["Real", "Fraud"])
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "confusion_matrix_heatmap.png"), dpi=300)
plt.show()


# =========================
# 13. ROC Curve
# =========================

fpr, tpr, thresholds = roc_curve(val_true, val_probs)

plt.figure(figsize=(6, 5))
plt.plot(fpr, tpr, label=f"AUC = {auc:.4f}")
plt.plot([0, 1], [0, 1], linestyle="--")
plt.title("ROC Curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "roc_curve.png"), dpi=300)
plt.show()


# =========================
# 14. APCER at 1% BPCER
# =========================

def calculate_apcer_at_bpcer(y_true, scores, bpcer_target=0.01):
    y_true = np.array(y_true)
    scores = np.array(scores)

    bona_fide_scores = scores[y_true == 0]
    attack_scores = scores[y_true == 1]

    threshold = np.quantile(bona_fide_scores, 1 - bpcer_target)

    bpcer = np.mean(bona_fide_scores >= threshold)
    apcer = np.mean(attack_scores < threshold)

    return {
        "threshold_at_1_percent_bpcer": float(threshold),
        "bpcer": float(bpcer),
        "apcer": float(apcer)
    }

det_metrics = calculate_apcer_at_bpcer(val_true, val_probs)

with open(os.path.join(RESULT_DIR, PREFIX + "det_apcer_bpcer_metrics.json"), "w") as f:
    json.dump(det_metrics, f, indent=4)

print("APCER/BPCER metrics:", det_metrics)


# =========================
# 15. Save Validation Predictions
# =========================

val_result_df = val_data[["id", "label", "full_path"]].copy()
val_result_df["predicted_score"] = val_probs
val_result_df["predicted_label"] = val_pred

val_result_df.to_csv(
    os.path.join(RESULT_DIR, PREFIX + "validation_predictions.csv"),
    index=False
)

plt.figure(figsize=(7, 5))
plt.hist(val_probs, bins=30)
plt.title("Validation Prediction Score Distribution")
plt.xlabel("Predicted Fraud Score")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "prediction_score_distribution.png"), dpi=300)
plt.show()


# =========================
# 16. Prepare Test Paths
# =========================

def find_test_image(sample_id):
    possible_paths = [
        os.path.join(TEST_IMG_DIR, sample_id + ".jpg"),
        os.path.join(TEST_IMG_DIR, sample_id + ".png"),
        os.path.join(TEST_IMG_DIR, sample_id + ".jpeg"),
        os.path.join(TEST_IMG_DIR, sample_id + ".webp"),
    ]

    for p in possible_paths:
        if os.path.exists(p):
            return p

    # fallback search
    for root, dirs, files in os.walk(TEST_IMG_DIR):
        for file in files:
            name, ext = os.path.splitext(file)
            if name == sample_id:
                return os.path.join(root, file)

    return None


sub_df["full_path"] = sub_df["id"].apply(find_test_image)
missing_test = sub_df["full_path"].isna().sum()

print("Missing test images:", missing_test)

test_df = sub_df.dropna(subset=["full_path"]).reset_index(drop=True)


# =========================
# 17. Test Dataset
# =========================

def make_test_dataset(df):
    paths = df["full_path"].values
    ds = tf.data.Dataset.from_tensor_slices(paths)
    ds = ds.map(lambda x: load_image(x, None), num_parallel_calls=tf.data.AUTOTUNE)
    ds = ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
    return ds

test_ds = make_test_dataset(test_df)


# =========================
# 18. Predict Test Scores
# =========================

test_scores = model.predict(test_ds).ravel()

test_df["label"] = test_scores

final_submission = sub_df[["id"]].merge(
    test_df[["id", "label"]],
    on="id",
    how="left"
)

final_submission["label"] = final_submission["label"].fillna(0.0)

submission_path = os.path.join(RESULT_DIR, PREFIX + "submission.csv")
final_submission.to_csv(submission_path, index=False)

print("Submission saved:", submission_path)
print(final_submission.head())


# =========================
# 19. Test Prediction Graph
# =========================

plt.figure(figsize=(7, 5))
plt.hist(final_submission["label"], bins=40)
plt.title("Public Test Fraud Score Distribution")
plt.xlabel("Fraud Score")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "test_prediction_distribution.png"), dpi=300)
plt.show()


# =========================
# 20. Save Summary TXT
# =========================

summary_path = os.path.join(RESULT_DIR, PREFIX + "result_summary.txt")

with open(summary_path, "w") as f:
    f.write("FREUID Fraud Prediction Results\n")
    f.write("===============================\n\n")
    f.write(f"Total training images used: {len(full_df)}\n")
    f.write(f"Validation accuracy: {acc:.4f}\n")
    f.write(f"Validation ROC AUC: {auc:.4f}\n")
    f.write(f"APCER at 1% BPCER: {det_metrics['apcer']:.4f}\n")
    f.write(f"BPCER: {det_metrics['bpcer']:.4f}\n")
    f.write(f"Threshold: {det_metrics['threshold_at_1_percent_bpcer']:.4f}\n\n")
    f.write("Saved files:\n")
    f.write("- fraud_best_model.keras\n")
    f.write("- fraud_final_model.keras\n")
    f.write("- fraud_final_model.h5\n")
    f.write("- fraud_training_history.csv\n")
    f.write("- fraud_metrics.json\n")
    f.write("- fraud_classification_report.csv\n")
    f.write("- fraud_confusion_matrix_heatmap.png\n")
    f.write("- fraud_roc_curve.png\n")
    f.write("- fraud_submission.csv\n")

print("All results saved successfully in:", RESULT_DIR)
```
