```python
import os
import json
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, classification_report, confusion_matrix,
    roc_auc_score, roc_curve
)
from sklearn.utils.class_weight import compute_class_weight

import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau


# =========================
# 1. Paths
# =========================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Fraud Prediction"
DATA_DIR = os.path.join(BASE_DIR, "the-freuid-challenge-2026-ijcai-ecai")

TRAIN_IMG_DIR = os.path.join(DATA_DIR, "train", "train")
TRAIN_SAMPLE_IMG_DIR = os.path.join(DATA_DIR, "train_sample", "train_sample")
TEST_IMG_DIR = os.path.join(DATA_DIR, "public_test", "public_test")

TRAIN_LABELS = os.path.join(DATA_DIR, "train_labels.csv")
TRAIN_SAMPLE_LABELS = os.path.join(DATA_DIR, "train_sample_labels.csv")
SAMPLE_SUBMISSION = os.path.join(DATA_DIR, "sample_submission.csv")

RESULT_DIR = os.path.join(BASE_DIR, "fraud_results_class_weighted")
os.makedirs(RESULT_DIR, exist_ok=True)

PREFIX = "fraud_weighted_"

print("Results will be saved in:", RESULT_DIR)


# =========================
# 2. Settings
# =========================

IMG_SIZE = 224
BATCH_SIZE = 32
EPOCHS = 15
SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)


# =========================
# 3. Load CSV Files
# =========================

train_df = pd.read_csv(TRAIN_LABELS)
sample_df = pd.read_csv(TRAIN_SAMPLE_LABELS)
submission_df = pd.read_csv(SAMPLE_SUBMISSION)

print("Train labels:", train_df.shape)
print("Train sample labels:", sample_df.shape)
print("Sample submission:", submission_df.shape)

print(train_df.head())


# =========================
# 4. Auto-detect Columns
# =========================

def find_id_column(df):
    possible = ["id", "image_id", "filename", "file_name", "image", "path"]
    for col in possible:
        if col in df.columns:
            return col
    return df.columns[0]


def find_label_column(df):
    possible = ["label", "target", "class", "is_fraud", "fraud"]
    for col in possible:
        if col in df.columns:
            return col
    return df.columns[-1]


ID_COL = find_id_column(train_df)
LABEL_COL = find_label_column(train_df)

print("ID column:", ID_COL)
print("Label column:", LABEL_COL)


# =========================
# 5. Resolve Image Paths
# =========================

def find_image_path(image_id, folder):
    image_id = str(image_id)

    possible_names = [
        image_id,
        image_id + ".jpg",
        image_id + ".jpeg",
        image_id + ".png",
        image_id + ".webp"
    ]

    for name in possible_names:
        path = os.path.join(folder, name)
        if os.path.exists(path):
            return path

    for root, dirs, files in os.walk(folder):
        for file in files:
            file_name, ext = os.path.splitext(file)
            if file_name == image_id or file == image_id:
                return os.path.join(root, file)

    return None


train_df["full_path"] = train_df[ID_COL].apply(lambda x: find_image_path(x, TRAIN_IMG_DIR))
sample_df["full_path"] = sample_df[ID_COL].apply(lambda x: find_image_path(x, TRAIN_SAMPLE_IMG_DIR))

train_df = train_df.dropna(subset=["full_path"]).reset_index(drop=True)
sample_df = sample_df.dropna(subset=["full_path"]).reset_index(drop=True)

print("Valid train images:", train_df.shape)
print("Valid sample images:", sample_df.shape)

full_df = pd.concat([train_df, sample_df], ignore_index=True)
full_df["label"] = full_df[LABEL_COL].astype(int)

full_df.to_csv(os.path.join(RESULT_DIR, PREFIX + "final_training_data.csv"), index=False)


# =========================
# 6. Label Distribution
# =========================

label_counts = full_df["label"].value_counts().sort_index()
print(label_counts)

plt.figure(figsize=(6, 4))
label_counts.plot(kind="bar")
plt.title("Fraud Label Distribution")
plt.xlabel("Label")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "label_distribution_graph.png"), dpi=300)
plt.show()


# =========================
# 7. Class Weights
# =========================

classes = np.unique(full_df["label"])

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=full_df["label"]
)

class_weights = {
    int(classes[i]): float(weights[i])
    for i in range(len(classes))
}

print("Auto class weights:", class_weights)

# Extra fraud boost
# If label 1 means fraud, this increases fraud importance more.
class_weights[1] = class_weights[1] * 2.0

print("Final class weights after fraud boost:", class_weights)

with open(os.path.join(RESULT_DIR, PREFIX + "class_weights.json"), "w") as f:
    json.dump(class_weights, f, indent=4)


# =========================
# 8. Train Validation Split
# =========================

train_data, val_data = train_test_split(
    full_df,
    test_size=0.2,
    random_state=SEED,
    stratify=full_df["label"]
)

print("Training data:", train_data.shape)
print("Validation data:", val_data.shape)


# =========================
# 9. TensorFlow Dataset
# =========================

def load_image(path, label):
    image = tf.io.read_file(path)
    image = tf.image.decode_image(image, channels=3, expand_animations=False)
    image = tf.image.resize(image, [IMG_SIZE, IMG_SIZE])
    image = image / 255.0

    return image, tf.cast(label, tf.float32)


def make_dataset(df, shuffle=True):
    paths = df["full_path"].values
    labels = df["label"].values

    ds = tf.data.Dataset.from_tensor_slices((paths, labels))
    ds = ds.map(load_image, num_parallel_calls=tf.data.AUTOTUNE)

    if shuffle:
        ds = ds.shuffle(1000, seed=SEED)

    ds = ds.batch(BATCH_SIZE)
    ds = ds.prefetch(tf.data.AUTOTUNE)

    return ds


train_ds = make_dataset(train_data, shuffle=True)
val_ds = make_dataset(val_data, shuffle=False)


# =========================
# 10. Build Model
# =========================

data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.05),
    layers.RandomZoom(0.10),
    layers.RandomContrast(0.10),
])

base_model = tf.keras.applications.EfficientNetB0(
    include_top=False,
    weights="imagenet",
    input_shape=(IMG_SIZE, IMG_SIZE, 3)
)

base_model.trainable = False

inputs = layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3))
x = data_augmentation(inputs)
x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.35)(x)
x = layers.Dense(128, activation="relu")(x)
x = layers.Dropout(0.35)(x)
outputs = layers.Dense(1, activation="sigmoid")(x)

model = models.Model(inputs, outputs)

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0005),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.AUC(name="auc"),
        tf.keras.metrics.Recall(name="recall"),
        tf.keras.metrics.Precision(name="precision")
    ]
)

model.summary()


# =========================
# 11. Train Model with Class Weights
# =========================

best_model_path = os.path.join(RESULT_DIR, PREFIX + "best_model.keras")

callbacks = [
    EarlyStopping(
        monitor="val_auc",
        patience=4,
        mode="max",
        restore_best_weights=True
    ),
    ModelCheckpoint(
        best_model_path,
        monitor="val_auc",
        mode="max",
        save_best_only=True
    ),
    ReduceLROnPlateau(
        monitor="val_loss",
        patience=2,
        factor=0.3,
        verbose=1
    )
]

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    callbacks=callbacks,
    class_weight=class_weights
)

model.save(os.path.join(RESULT_DIR, PREFIX + "final_model.keras"))
model.save(os.path.join(RESULT_DIR, PREFIX + "final_model.h5"))


# =========================
# 12. Training Graphs
# =========================

history_df = pd.DataFrame(history.history)
history_df.to_csv(os.path.join(RESULT_DIR, PREFIX + "training_history.csv"), index=False)

plt.figure(figsize=(7, 5))
plt.plot(history.history["accuracy"], label="Train Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.title("Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "training_accuracy_graph.png"), dpi=300)
plt.show()

plt.figure(figsize=(7, 5))
plt.plot(history.history["loss"], label="Train Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "training_loss_graph.png"), dpi=300)
plt.show()

plt.figure(figsize=(7, 5))
plt.plot(history.history["auc"], label="Train AUC")
plt.plot(history.history["val_auc"], label="Validation AUC")
plt.title("AUC Graph")
plt.xlabel("Epoch")
plt.ylabel("AUC")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "auc_graph.png"), dpi=300)
plt.show()

plt.figure(figsize=(7, 5))
plt.plot(history.history["recall"], label="Train Recall")
plt.plot(history.history["val_recall"], label="Validation Recall")
plt.title("Fraud Recall Graph")
plt.xlabel("Epoch")
plt.ylabel("Recall")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "fraud_recall_graph.png"), dpi=300)
plt.show()


# =========================
# 13. Validation Prediction
# =========================

val_probs = model.predict(val_ds).ravel()
val_true = val_data["label"].values

# Lower threshold = detects more fraud
THRESHOLD = 0.30

val_pred = (val_probs >= THRESHOLD).astype(int)

accuracy = accuracy_score(val_true, val_pred)
auc_score = roc_auc_score(val_true, val_probs)
cm = confusion_matrix(val_true, val_pred)
report = classification_report(val_true, val_pred, output_dict=True)

print("Threshold used:", THRESHOLD)
print("Accuracy:", accuracy)
print("ROC AUC:", auc_score)
print(classification_report(val_true, val_pred))

metrics = {
    "threshold": THRESHOLD,
    "accuracy": float(accuracy),
    "roc_auc": float(auc_score),
    "confusion_matrix": cm.tolist(),
    "classification_report": report,
    "class_weights": class_weights
}

with open(os.path.join(RESULT_DIR, PREFIX + "metrics.json"), "w") as f:
    json.dump(metrics, f, indent=4)

pd.DataFrame(report).transpose().to_csv(
    os.path.join(RESULT_DIR, PREFIX + "classification_report.csv")
)


# =========================
# 14. Confusion Matrix Heatmap
# =========================

plt.figure(figsize=(5, 4))
plt.imshow(cm)
plt.title("Confusion Matrix Heatmap")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.colorbar()

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j], ha="center", va="center")

plt.xticks([0, 1], ["Real", "Fraud"])
plt.yticks([0, 1], ["Real", "Fraud"])
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "confusion_matrix_heatmap.png"), dpi=300)
plt.show()


# =========================
# 15. ROC Curve
# =========================

fpr, tpr, thresholds = roc_curve(val_true, val_probs)

plt.figure(figsize=(6, 5))
plt.plot(fpr, tpr, label=f"AUC = {auc_score:.4f}")
plt.plot([0, 1], [0, 1], linestyle="--")
plt.title("ROC Curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "roc_curve.png"), dpi=300)
plt.show()


# =========================
# 16. Validation Predictions CSV
# =========================

val_result_df = val_data[[ID_COL, "label", "full_path"]].copy()
val_result_df["fraud_score"] = val_probs
val_result_df["predicted_label"] = val_pred
val_result_df["threshold"] = THRESHOLD

val_result_df.to_csv(
    os.path.join(RESULT_DIR, PREFIX + "validation_predictions.csv"),
    index=False
)

plt.figure(figsize=(7, 5))
plt.hist(val_probs, bins=30)
plt.title("Validation Fraud Score Distribution")
plt.xlabel("Fraud Score")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "validation_score_distribution.png"), dpi=300)
plt.show()


# =========================
# 17. Test Image Paths
# =========================

SUB_ID_COL = find_id_column(submission_df)

def find_test_image(sample_id):
    return find_image_path(sample_id, TEST_IMG_DIR)

submission_df["full_path"] = submission_df[SUB_ID_COL].apply(find_test_image)

print("Missing test images:", submission_df["full_path"].isna().sum())

test_df = submission_df.dropna(subset=["full_path"]).reset_index(drop=True)


# =========================
# 18. Test Dataset
# =========================

def load_test_image(path):
    image = tf.io.read_file(path)
    image = tf.image.decode_image(image, channels=3, expand_animations=False)
    image = tf.image.resize(image, [IMG_SIZE, IMG_SIZE])
    image = image / 255.0
    return image


def make_test_dataset(df):
    paths = df["full_path"].values
    ds = tf.data.Dataset.from_tensor_slices(paths)
    ds = ds.map(load_test_image, num_parallel_calls=tf.data.AUTOTUNE)
    ds = ds.batch(BATCH_SIZE)
    ds = ds.prefetch(tf.data.AUTOTUNE)
    return ds


test_ds = make_test_dataset(test_df)


# =========================
# 19. Predict Test Fraud Scores
# =========================

test_scores = model.predict(test_ds).ravel()

test_df["label"] = test_scores

final_submission = submission_df[[SUB_ID_COL]].merge(
    test_df[[SUB_ID_COL, "label"]],
    on=SUB_ID_COL,
    how="left"
)

final_submission["label"] = final_submission["label"].fillna(0.0)

submission_path = os.path.join(RESULT_DIR, PREFIX + "submission.csv")
final_submission.to_csv(submission_path, index=False)

print("Submission saved at:", submission_path)
print(final_submission.head())


# =========================
# 20. Test Score Graph
# =========================

plt.figure(figsize=(7, 5))
plt.hist(final_submission["label"], bins=40)
plt.title("Public Test Fraud Score Distribution")
plt.xlabel("Fraud Score")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, PREFIX + "test_score_distribution.png"), dpi=300)
plt.show()


# =========================
# 21. Final Summary
# =========================

summary_path = os.path.join(RESULT_DIR, PREFIX + "result_summary.txt")

with open(summary_path, "w") as f:
    f.write("FREUID Fraud Detection with Class Weights\n")
    f.write("========================================\n\n")
    f.write(f"Total images used: {len(full_df)}\n")
    f.write(f"Threshold used: {THRESHOLD}\n")
    f.write(f"Accuracy: {accuracy:.4f}\n")
    f.write(f"ROC AUC: {auc_score:.4f}\n")
    f.write(f"Class weights: {class_weights}\n\n")
    f.write("Important Note:\n")
    f.write("Class weights were used to give more importance to fraud samples.\n")
    f.write("Threshold was lowered to 0.30 to detect fraud more aggressively.\n\n")
    f.write("Generated Files:\n")
    f.write("- fraud_weighted_best_model.keras\n")
    f.write("- fraud_weighted_final_model.keras\n")
    f.write("- fraud_weighted_final_model.h5\n")
    f.write("- fraud_weighted_submission.csv\n")
    f.write("- fraud_weighted_metrics.json\n")
    f.write("- fraud_weighted_classification_report.csv\n")
    f.write("- fraud_weighted_confusion_matrix_heatmap.png\n")
    f.write("- fraud_weighted_roc_curve.png\n")
    f.write("- fraud_weighted_fraud_recall_graph.png\n")

print("All class-weighted fraud detection results saved in:", RESULT_DIR)
```
